```python
numerical_features = [
    "Age",
    "Income",
    "Credit_Score",
    "Credit_Utilization",
    "Missed_Payments",
    "Loan_Balance",
    "Debt_to_Income_Ratio",
    "Account_Tenure"
]

for column in numerical_features:
    plt.figure(figsize=(8, 4))
    sns.histplot(data=df, x=column, kde=True, bins=30)
    plt.title(f"Distribution of {column}")
    plt.show()
```

```python
features_to_compare = [
    "Credit_Utilization",
    "Credit_Score",
    "Missed_Payments",
    "Debt_to_Income_Ratio",
    "Income",
    "Loan_Balance",
    "Age",
    "Account_Tenure"
]

for column in features_to_compare:
    plt.figure(figsize=(7, 4))

    sns.boxplot(
        data=df,
        x="Delinquent_Account",
        y=column
    )

    plt.title(f"{column} vs Delinquency")
    plt.xlabel("Delinquent Account (0 = No, 1 = Yes)")
    plt.show()
```

```python
correlation = df.select_dtypes(include=np.number).corr()
plt.figure(figsize=(13, 9))
sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)
plt.title("CreditGuard - Correlation Heatmap")
plt.show()
```

```python
payment_columns = [
    "Month_1",
    "Month_2",
    "Month_3",
    "Month_4",
    "Month_5",
    "Month_6"
]

for column in payment_columns:
    print(f"\n{column}")
    print(df[column].value_counts(normalize=True).mul(100).round(2))
```

```python
payment_summary = pd.DataFrame({
    column: df[column].value_counts(normalize=True) * 100
    for column in payment_columns
}).T

payment_summary.plot(
    kind="bar",
    figsize=(12, 6),
    stacked=True
)
plt.title("Monthly Payment Behaviour")
plt.xlabel("Month")
plt.ylabel("Percentage of Customers")
plt.legend(title="Payment Status")
plt.xticks(rotation=0)
plt.show()
```

```python
categorical_features = [
    "Employment_Status",
    "Credit_Card_Type",
    "Location"
]

for column in categorical_features:

    delinquency_rate = (
        df.groupby(column, dropna=False)["Delinquent_Account"]
        .mean() * 100
    ).sort_values(ascending=False)

    print(f"\nDelinquency Rate by {column}")
    print(delinquency_rate.round(2))

    delinquency_rate.plot(
        kind="bar",
        figsize=(9, 4)
    )

    plt.title(f"Delinquency Rate by {column}")
    plt.ylabel("Delinquency Rate (%)")
    plt.xlabel(column)
    plt.xticks(rotation=35)
    plt.show()
```


# CreditGuard - EDA Findings

## 1. Dataset Overview
- 500 customer records and 19 columns.
- No duplicate rows or duplicate Customer_ID values.
- Target variable: Delinquent_Account.

## 2. Data Quality Findings
- Income: 39 missing values (7.8%).
- Loan_Balance: 29 missing values (5.8%).
- Credit_Score: 2 missing values (0.4%).
- Employment_Status contains inconsistent labels and casing.
- Credit_Utilization requires unit and range verification.
- Payment history contains text labels that need encoding verification.

## 3. Target Distribution
- Non-delinquent accounts: 420 (84%).
- Delinquent accounts: 80 (16%).
- The target variable is imbalanced.

## 4. Initial Observations
- Credit-card type and location show differences in observed delinquency rates.
- Monthly payment-history columns contain useful behavioural information to investigate.
- Missed_Payments has an unexpected relationship with the target and requires further investigation.
- No causal conclusions or confirmed feature importance can be established from EDA alone.

## 5. Limitations
- Dataset size is relatively small.
- Missing values and inconsistent categories require investigation.
- Prediction timing must be established to prevent data leakage.
- Fairness and model reliability require further assessment.

## 6. Conclusion
EDA identified data-quality issues, class imbalance and potential patterns worth investigating. These findings will guide the next stage of CreditGuard: data preprocessing and feature preparation.
